## 4. Descrevam as informações contidas nos dados

Expliquem o significado dos registros antes de calcular estatísticas.

- **O que uma linha representa?** 
Cada linha é um curso de uma instituição, em uma cidade. No presencial, cada curso ocupa uma única linha, no município da oferta. No ensino a distância, o mesmo curso pode ocupar várias linhas: uma por município em que há alunos e outra, só no nível Brasil, para vagas e inscritos, que o Inep não reparte por local.


- **Como os dados foram produzidos?**
Segundo o site do Inep: "A coleta utiliza informações do cadastro do Sistema e-MEC, que reúne registros das instituições, dos cursos e dos locais de oferta."

- **Qual é a cobertura temporal e geográfica?** 
Neste trabalho vou utilizar dois anos do censo: 2023 e 2024. Os dados são de instituições de ensino superior em todo o Brasil.

- **Quem ou o que pode estar ausente da base?** 
Cursos fora do Brasil e cursos que não são de graduação. Informações sobre cada aluno individualmente também não estão disponíveis. Apenas os números de anlunos em cada curso.

- **Há mais de uma tabela? Como elas se relacionam?** 
Sim. Em cada ano há duas tabelas: `MICRODADOS_CADASTRO_CURSOS`, com um curso por linha, e `MICRODADOS_ED_SUP_IES`, com uma instituição por linha (nome, sigla e dados de docentes). Vários cursos se ligam a uma instituição por `CO_IES`. Os dois anos são arquivos separados e se pareiam pela oferta: `CO_IES`, `CO_CURSO`, `CO_MUNICIPIO` e `TP_DIMENSAO`. O dicionário oficial está na planilha anexa, nas abas `cadastro_cursos` e `cadastro_ies`.

### Dicionário dos principais atributos


| Atributo | Significado | Tipo semântico | Unidade ou categorias | Observações |
|---|---|---|---|---|
| `NU_ANO_CENSO` | Ano de referência do censo | temporal | 2023 ou 2024 | Cada CSV contém um único ano |
| `CO_IES` | Código da instituição | identificador | Código numérico do Inep | Liga o curso ao arquivo de IES, onde estão `NO_IES` e `SG_IES` |
| `CO_CURSO` | Código do curso | identificador | Código numérico | Identifica o curso dentro da instituição |
| `NO_CURSO` | Nome do curso | categórico | Texto, como Física ou Medicina | O mesmo nome ocorre em várias instituições |
| `TP_MODALIDADE_ENSINO` | Modalidade de ensino | categórico | 1 presencial; 2 a distância | O recorte da proposta usa o código 1 |
| `TP_NIVEL_ACADEMICO` | Nível do curso | categórico | 1 graduação; 2 sequencial de formação específica | Quase todas as linhas são graduação |
| `TP_GRAU_ACADEMICO` | Grau conferido na conclusão | categórico | 1 bacharelado; 2 licenciatura; 3 tecnológico; 4 bacharelado e licenciatura | Em branco quando não se aplica, como na área básica de ingresso |
| `TP_REDE` | Rede de ensino | categórico | 1 pública; 2 privada | |
| `TP_CATEGORIA_ADMINISTRATIVA` | Categoria administrativa da instituição | categórico | 1 federal; 2 estadual; 3 municipal; 4 privada com fins lucrativos; 5 privada sem fins lucrativos; 7 especial | |
| `TP_DIMENSAO` | Tipo de linha geográfica | categórico | 1 presencial no Brasil; 2 EaD por município; 3 EaD só no nível Brasil; 4 EaD no exterior | Em 2024, matrículas estão nas dimensões 1, 2 e 4; vagas e inscritos da EaD estão na dimensão 3 |
| `SG_UF` | Sigla da UF do local de oferta | geográfico | 27 unidades da federação | Ausente na EaD agregada no Brasil e no exterior |
| `NO_MUNICIPIO` | Município do local de oferta | geográfico | Nome do município | Ausente nas mesmas linhas sem UF |
| `NO_CINE_AREA_GERAL` | Área geral do curso | categórico | Áreas da classificação CINE/Unesco adaptada, como Saúde e bem-estar, Educação e Engenharia, produção e construção | Há 11 áreas na base de 2024 |
| `QT_MAT` | Quantidade de matrículas | numérico | Alunos | Alunos com vínculo cursando ou formado no ano |
| `QT_ING` | Quantidade de ingressantes | numérico | Alunos | Ingresso entre 1º de janeiro e 1º de julho do ano do censo |
| `QT_CONC` | Quantidade de concluintes | numérico | Alunos | Concluíram o curso no ano de referência |
